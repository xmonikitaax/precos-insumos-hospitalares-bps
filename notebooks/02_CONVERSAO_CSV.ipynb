## 3. Configuração
> ⚙️ Ajuste os dois caminhos abaixo.

```python
import gc, pandas as pd
from pathlib import Path

PASTA_ORIGINAIS  = Path(PASTA_DADOS) / 'Base Tratamento Minimo'
PASTA_CONVERSAO  = Path(PASTA_DADOS) / 'Base Conversao CSV'

Path(PASTA_CONVERSAO).mkdir(parents=True, exist_ok=True)

def detectar_header(caminho):
    """Detecta a linha do cabeçalho real (arquivos antigos têm aviso na linha 0)."""
    df_raw = pd.read_excel(caminho, header=None, nrows=5)
    for i, row in df_raw.iterrows():
        if sum(1 for v in row.values if pd.notna(v)) >= 3:
            return i
    return 0

def converter(nome_arquivo):
    """Converte um arquivo Excel BPS para CSV. Pula se já existir."""
    origem = Path(PASTA_ORIGINAIS) / nome_arquivo
    destino = Path(PASTA_CONVERSAO) / (Path(nome_arquivo).stem + '.csv')

    if destino.exists():
        print(f'  {nome_arquivo} → já convertido, pulando ✓')
        return

    if not origem.exists():
        print(f'  {nome_arquivo} → NÃO ENCONTRADO em {PASTA_ORIGINAIS}')
        return

    print(f'  Lendo {nome_arquivo}...', end=' ')
    hr = detectar_header(str(origem))
    df = pd.read_excel(str(origem), header=hr)
    linhas = len(df)
    df.to_csv(str(destino), index=False, encoding='utf-8-sig')
    mb = destino.stat().st_size / (1024 * 1024)
    print(f'{linhas:,} linhas → {destino.name} ({mb:.1f} MB) ✓')
    del df
    gc.collect()

print('Configuração OK ✓')
print(f'Origem : {PASTA_ORIGINAIS}')
print(f'Destino: {PASTA_CONVERSAO}')
```
